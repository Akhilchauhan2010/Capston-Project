```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

print("Starting the scrape... Please wait!")
book_data = []

for page in range(1, 51):
    url = f"https://books.toscrape.com/catalogue/page-{page}.html"
    response = requests.get(url)
    
    
    if response.status_code != 200:
        print(f"Uh oh! Failed to load page {page}")
        continue
        
    soup = BeautifulSoup(response.text, 'html.parser')
    
    books = soup.find_all('article', class_='product_pod')
    
    for book in books:
        
        title = book.h3.a['title']
        
        price_str = book.find('p', class_='price_color').text
        price = price_str.replace('£', '').replace('Â', '') 
        
        rating_class = book.p['class'][1] 
       
        availability = book.find('p', class_='instock availability').text.strip()
        
        book_data.append({
            'Title': title,
            'Price_GBP': price,
            'Rating_Text': rating_class,
            'Availability': availability
        })
        
    print(f"Scraped page {page}/50...")
    time.sleep(1) 

# Save raw dataset
df_raw = pd.DataFrame(book_data)
df_raw.to_csv('raw_books_data.csv', index=False, encoding='utf-8-sig')
print(f"Success! Saved {len(df_raw)} books to raw_books_data.csv")
```
