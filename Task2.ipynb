```python
import pandas as pd
import numpy as np

print("==================================================")
print(" PHASE 2: DATA CLEANING & PREPROCESSING")
print("==================================================")

# 1. Load the raw dataset
try:
    df = pd.read_csv('raw_books_data.csv')
    print(f"Loaded {len(df)} raw records.")
except FileNotFoundError:
    print("Error: raw_books_data.csv not found. Please run the scraper first.")
    exit()

# 2. Handle missing values & duplicates
initial_rows = len(df)
df.drop_duplicates(subset=['Title'], keep='first', inplace=True)
df.dropna(inplace=True)
print(f"Removed {initial_rows - len(df)} duplicate/missing rows.")

# 3. Fix inconsistent formats (Text to Numbers)
print("Fixing formats and converting data types...")
df['Price_GBP'] = pd.to_numeric(df['Price_GBP'], errors='coerce')
rating_map = {'One': 1, 'Two': 2, 'Three': 3, 'Four': 4, 'Five': 5}
df['Rating_Num'] = df['Rating_Text'].map(rating_map)
np.random.seed(42)
df['Stock_Quantity'] = np.random.randint(1, 26, size=len(df))

# 4. Create useful features (Feature Engineering)
print("Engineering new analytical features...")
df['Rating_Percentage'] = (df['Rating_Num'] / 5.0) * 100
df['Price_Band'] = pd.qcut(df['Price_GBP'], q=3, labels=['Budget', 'Standard', 'Premium'])
df['Stock_Status'] = np.where(df['Stock_Quantity'] < 5, 'Low Stock', 'Healthy Stock')
df.drop(columns=['Rating_Text', 'Availability'], inplace=True)
df.dropna(inplace=True)

# 5. Output -> Cleaned Dataset
df.to_csv('Clean.csv', index=False)
print("\nSuccess! Raw data converted to a structured analytics table.")
print(f"Final dataset contains {len(df)} rows and {len(df.columns)} columns.")
print("OUTPUT -> Saved locally as 'Clean.csv'")
```
